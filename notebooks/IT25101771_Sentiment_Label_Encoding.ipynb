## 1. Technique Explanation

**Encoding categorical variables** converts non-numeric category labels into a numeric form
that machine learning algorithms can process. Two common forms are demonstrated here:
**label encoding** (mapping each category to a single integer) and **one-hot encoding**
(representing each category as a binary vector).

## 2. Justification for This Dataset

Our target variable is sentiment: conceptually the categories "positive" and "negative." The
dataset already provides this as integer labels (1/0), but it is important to demonstrate *why*
that representation exists and *how* it is produced, since some model training setups (like
categorical cross-entropy loss) expect a one-hot encoded form rather than a single integer.
This also documents that our encoding is a deliberate, understood choice — not an
unexplained given.


## 3. Implementation

```python
from sklearn.preprocessing import LabelEncoder

# Recreate the categorical form to demonstrate the encoding step explicitly
train_df["sentiment_category"] = train_df["label"].map({1: "positive", 0: "negative"})

# Label encoding: category -> single integer
le = LabelEncoder()
train_df["label_encoded"] = le.fit_transform(train_df["sentiment_category"])

print("Classes learned by the encoder:", list(le.classes_))
train_df[["sentiment_category", "label_encoded", "label"]].sample(5, random_state=1)
```

```python
# One-hot encoding: category -> binary vector (useful for categorical cross-entropy setups)
one_hot = pd.get_dummies(train_df["sentiment_category"], prefix="sentiment")
one_hot.sample(5, random_state=1)
```

## 4. EDA Visualization & Interpretation

A count plot of the encoded labels doubles as our class-balance check.


```python
plt.figure(figsize=(5, 4))
sns.countplot(data=train_df, x="sentiment_category", order=["negative", "positive"],
              palette={"negative": "#C1272D", "positive": "#2E9E6D"})
plt.title("Class Distribution After Encoding")
plt.xlabel("")
plt.ylabel("Count")
plt.show()

train_df["sentiment_category"].value_counts()
```
